El equipo de analistas necesita saber si el pago mensual se comporta como una variable normalmente distribuida, ya que algunas decisiones estadísticas y operativas requieren este supuesto.

">base de datos aquí</a>** y responde las siguientes preguntas:

1. Calcula la media, la mediana y la moda de la variable Pago mensual. ¿Los valores obtenidos son similares entre sí? Analiza lo que esto indica sobre la simetría de la distribución de los datos.

```python
import pandas as pd

df_1=pd.read_excel("/content/Base8_1.xlsx")

df_1.head()
```

```python
media=df_1["Pago mensual ($)"].median()
moda=df_1["Pago mensual ($)"].mode()
mediana=df_1["Pago mensual ($)"].median()

print(f"Media: {media}, Mediana: {mediana}")

if len(moda)==1:
  print(f'Moda: ${moda.to_list()[0]}')
else:
  print(f"Moda: ${moda.to_list()}")

minimo=df_1["Pago mensual ($)"].min()
maximo=df_1["Pago mensual ($)"].max()

print(f"\n Minimo: ${minimo}\n Maximo: ${maximo}")
```

2. Calcula e interpreta el coeficiente de asimetría y el coeficiente de curtosis para la variable Pago mensual. ¿Estos valores son consistentes con los esperados en una distribución normal? Justifica tu respuesta.

```python
from scipy.stats import skew, kurtosis

#Coeficiente de asimetria:
asimetria=skew(df_1["Pago mensual ($)"])

#Curtosis
curtosis=kurtosis(df_1["Pago mensual ($)"])

print(f"Coefiiciente de asimetria: {asimetria:.2f}")
print(f"Coeficiente de curtosis: {curtosis:.2f}")
```

4. Considerando que la variable Pago mensual sigue una distribución aproximadamente normal, responde las siguientes preguntas:

$\hspace{1cm}$ a)  ¿Cuál es la probabilidad de que un cliente pague menos de \$18.000 al mes?

$\hspace{1cm}$ b) ¿Cuál es la probabilidad de que un cliente pague más de \$25.000 al mes?

$\hspace{1cm}$ c) ¿Cuál es la probabilidad de que un cliente pague entre \$20.000 y \$24.000 al mes?

```python
#Calcular probabilidades:
import scipy.stats as stats

media_1=df_1["Pago mensual ($)"].mean()
desviacion_estandar=df_1["Pago mensual ($)"].std()

# Use stats.norm.cdf
p_menor_18000 = stats.norm.cdf(18000, loc=media_1, scale=desviacion_estandar) * 100

p_mayor_25000 = (1 - stats.norm.cdf(25000, loc=media_1, scale=desviacion_estandar)) * 100

p_entre_20_24 = (stats.norm.cdf(24000, loc=media_1, scale=desviacion_estandar) - stats.norm.cdf(20000, loc=media_1, scale=desviacion_estandar)) * 100

print(f"Probabilidad de que un cliente pague menos de $18.000 al mes: {p_menor_18000:.2f}%")
print(f"Probabilidad de que un cliente pague más de $25.000 al mes: {p_mayor_25000:.2f}%")
print(f"Probabilidad de que un cliente pague entre $20.000 y $24.000 al mes: {p_entre_20_24:.2f}%")
```
